```python
import statsmodels.formula.api as smf
import statsmodels.api as sm
```

```python
def vif_cal(Z):
    x_var = Z.copy()
    xvar_names = x_var.columns
    
    for i in range(len(xvar_names)):
        y_col = xvar_names[i]
        x_cols = xvar_names.drop(y_col)
        
        # build formula correctly
        formula = f"{y_col} ~ " + " + ".join(x_cols)
        
        model = smf.ols(formula=formula, data=x_var).fit()
        rsq = model.rsquared
        
        vif = round(1 / (1 - rsq), 2)
        print(y_col, "VIF =", vif)
```
